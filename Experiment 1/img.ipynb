```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
img = cv.imread(r'img.jpg')
print("Image Shape:", img.shape)
print("Image Data Type:", img.dtype)
```

```python
# Display the image using OpenCV
cv.imshow('image', img)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Display the image using Matplotlib
rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

plt.figure(figsize=(12,8))
plt.imshow(rgb)
plt.title("RGB Image")
plt.axis("off")
plt.show()
```

```python
resized = cv.resize(img, (850, 500))
cv.imshow('image', resized)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Access Pixel
print("Pixel Value at (594,963):")
print(img[594,963])
```

```python
#Pixel Modification
modified = img.copy()

modified[594,963] = [0,0,255]

cv.imshow("Modified Pixel", modified)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Add text
text = img.copy()

cv.putText(text,
            "Computer Vision Lab",
            (125,100),
            cv.FONT_HERSHEY_PLAIN,
            0.8,
            (0,0,0))

cv.imshow("Text",text)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Image Cropping
crop = img[200:800,450:1500]

cv.imshow("Cropped Image", crop)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
#BGR to Grayscale
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

cv.imshow("Grayscale Image", gray)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Image Transpose and Flip
transpose_flip = cv.transpose(img) 

cv.imshow("Transposed and Flipped Image", transpose_flip)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Draw Circle
circle = cv.circle(img.copy(),(447,253), 100, (0,0,0), -1)

cv.imshow("Circle", circle)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# Rotate Image
rotate = cv.rotate(img,cv.ROTATE_180)

cv.imshow("Rotated",rotate)
cv.waitKey(0)
cv.destroyAllWindows()
```
