```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread(r'nos.jpg')
plt.figure(figsize=(8, 4))
plt.imshow(img, cmap='gray')
plt.axis('off')
```

```python
#output size of the image = (height - kernel_size + 2 * padding) / stride + 1
#median filter
def median_filter(img, ksize =3 , stride = 1):
    height, width = img.shape[:2]
    pad = ksize // 2
    padded = np.pad(img, pad, mode='reflect')
    out_height = (height - ksize + 2 * pad) // stride + 1
    out_width = (width - ksize + 2 * pad) // stride + 1
    og = np.zeros((out_height, out_width), dtype=img.dtype)
    for i in range(out_height):
        for j in range(out_width):
                window = padded[i *stride: i*stride + ksize, j*stride : j*stride + ksize]
                og[i, j] = np.median(window)
                
    return og

#mean filter
def mean_filter(img, ksize =3 , stride = 1):
    height, width = img.shape[:2]
    pad = ksize // 2
    padded = np.pad(img, pad, mode='reflect')
    out_height = (height - ksize + 2 * pad) // stride + 1
    out_width = (width - ksize + 2 * pad) // stride + 1
    og = np.zeros((out_height, out_width), dtype=img.dtype)
    for i in range(out_height):
        for j in range(out_width):
                window = padded[i *stride: i*stride + ksize, j*stride : j*stride + ksize]
                og[i, j] = np.mean(window)
                
    return og
```

```python
clean_img_median = median_filter(img, ksize=3, stride=1)
clean_img_mean = mean_filter(img, ksize=3, stride=1)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(clean_img_median, cmap='gray')
plt.title('Median Filtered Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(clean_img_mean, cmap='gray')
plt.title('Mean Filtered Image')
plt.axis('off')
```

```python
k3 = median_filter(img, ksize=3, stride=1)
k5 = median_filter(img, ksize=5, stride=1)
k7 = median_filter(img, ksize=7, stride=1)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(k3, cmap="gray")
plt.title("Kernel 3x3")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(k5, cmap="gray")
plt.title("Kernel 5x5")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(k7, cmap="gray")
plt.title("Kernel 7x7")
plt.axis("off")

plt.show()
```

```python
s1 = median_filter(img, ksize=3, stride=1)
s2 = median_filter(img, ksize=3, stride=2)
s3 = median_filter(img, ksize=3, stride=3)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(s1, cmap="gray")
plt.title(f"Stride 1 {s1.shape}")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(s2, cmap="gray")
plt.title(f"Stride 2 {s2.shape}")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(s3, cmap="gray")
plt.title(f"Stride 3 {s3.shape}")
plt.axis("off")

plt.show()
```
