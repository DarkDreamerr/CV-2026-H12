```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread('img.jpg', cv.IMREAD_GRAYSCALE)
img.shape
plt.imshow(img, cmap='gray')
plt.axis('off')
```

```python
def salt_pepper_noise(image, prob):
    
    noisy = np.copy(image)
    
    random_matrix = np.random.rand(*image.shape)

    noisy[random_matrix < (prob / 2)] = 0
    noisy[random_matrix > 1 - (prob / 2)] = 255
    
    return noisy
```

```python
nos = salt_pepper_noise(img, 0.2)
plt.imshow(nos, cmap='gray')
plt.axis('off')
```

```python
# moving window
ksize = 3

move = nos.copy()
for i in range(move.shape[0]):
    for j in range(move.shape[1]):
        if i <= move.shape[0] - ksize and j <= move.shape[1] - ksize:
            window = nos[i :i + ksize, j : j + ksize]
            move[i, j] = np.median(window) 

plt.imshow(move, cmap='gray')
plt.axis('off')
```

```python
# median function
def median(values):
    flat = np.array(values).flatten()
    flat.sort()
    
    n = len(flat)
    mid = n // 2
    
    if n % 2 == 1:
        return flat[mid]
    else:
        return (flat[mid - 1] + flat[mid]) / 2
```

```python
window = [[12, 200, 15], [14, 18, 250], [11, 16, 13]]
print(median(window))
print(np.median(window))
```
